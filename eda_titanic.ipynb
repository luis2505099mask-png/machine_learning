# Análisis exploratorio del dataset Titanic
Cuaderno con pandas, seaborn y matplotlib sobre `titanic_clean.csv` (891 pasajeros, sin valores nulos). Cada punto lleva su explicación en una celda markdown.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

df = pd.read_csv("titanic_clean.csv")
df.head()
```

```python
df.info()
df.describe()
```

## 1. Distribución de edades de los pasajeros
Histograma con curva de densidad (KDE). La mayor parte de los pasajeros tiene entre 20 y 40 años. El pico alto cerca de los 28 años se debe a que en el dataset limpio las edades faltantes se rellenaron con un valor central (la mediana), lo que concentra muchos registros en ese valor.

```python
sns.histplot(df["Age"], bins=30, kde=True, color="steelblue")
plt.title("Distribución de edades de los pasajeros")
plt.xlabel("Edad")
plt.ylabel("Cantidad de pasajeros")
plt.show()
```

## 2. Sobrevivientes por clase de pasajero (Pclass)
Conteo de pasajeros que sobrevivieron y no sobrevivieron en cada clase. En números absolutos hay más sobrevivientes en 1ª clase (136) que en 3ª (119), a pesar de que la 3ª clase tenía más del doble de pasajeros: la clase alta tuvo una supervivencia mucho mayor (63 % vs. 24 %).

```python
sns.countplot(data=df, x="Pclass", hue="Survived", palette=["#d9534f", "#5cb85c"])
plt.title("Sobrevivientes por clase de pasajero")
plt.xlabel("Clase (Pclass)")
plt.ylabel("Cantidad de pasajeros")
plt.legend(title="Sobrevivió", labels=["No", "Sí"])
plt.show()
```

## 3. Sobrevivientes por género (Sex)
Las mujeres sobrevivieron en una proporción mucho mayor (74 %) que los hombres (19 %), consistente con la prioridad "mujeres y niños primero" en los botes salvavidas.

```python
sns.countplot(data=df, x="Sex", hue="Survived", palette=["#d9534f", "#5cb85c"])
plt.title("Sobrevivientes por género")
plt.xlabel("Género")
plt.ylabel("Cantidad de pasajeros")
plt.legend(title="Sobrevivió", labels=["No", "Sí"])
plt.show()
```

## 4. Sobrevivientes por puerto de embarque (Embarked)
Puertos: C = Cherbourg, Q = Queenstown, S = Southampton. Southampton tiene más sobrevivientes en números absolutos porque embarcó a la mayoría, pero Cherbourg tuvo la mejor tasa de supervivencia (55 %), probablemente porque allí subió una mayor proporción de pasajeros de 1ª clase.

```python
sns.countplot(data=df, x="Embarked", hue="Survived", order=["C", "Q", "S"], palette=["#d9534f", "#5cb85c"])
plt.title("Sobrevivientes por puerto de embarque")
plt.xlabel("Puerto de embarque")
plt.ylabel("Cantidad de pasajeros")
plt.legend(title="Sobrevivió", labels=["No", "Sí"])
plt.show()
```

## 5. Tasa de supervivencia por grupo de edad
Se agrupan las edades en categorías con `pd.cut`:

| Grupo | Rango |
|---|---|
| Infante | 0 – 5 |
| Niño | 6 – 12 |
| Adolescente | 13 – 18 |
| Joven Adulto | 19 – 35 |
| Adulto | 36 – 60 |
| Mayor | 61 en adelante |

Los infantes tienen la mayor tasa de supervivencia (≈ 70 %) y los mayores la menor (≈ 23 %). Los rangos son una decisión del análisis y pueden ajustarse en la lista `bins`.

```python
bins = [0, 5, 12, 18, 35, 60, 120]
labels = ["Infante", "Niño", "Adolescente", "Joven Adulto", "Adulto", "Mayor"]
df["GrupoEdad"] = pd.cut(df["Age"], bins=bins, labels=labels, include_lowest=True)

tasa_edad = df.groupby("GrupoEdad", observed=True)["Survived"].mean().reset_index()

sns.barplot(data=tasa_edad, x="GrupoEdad", y="Survived", hue="GrupoEdad", palette="viridis", legend=False)
plt.title("Tasa de supervivencia por grupo de edad")
plt.xlabel("Grupo de edad")
plt.ylabel("Tasa de supervivencia")
plt.ylim(0, 1)
plt.show()
```

## 6. Conteo de pasajeros por clase (Pclass)
La 3ª clase es la más numerosa (491 pasajeros), seguida de la 1ª (216) y la 2ª (184).

```python
sns.countplot(data=df, x="Pclass", hue="Pclass", palette="Blues", legend=False)
plt.title("Cantidad de pasajeros por clase")
plt.xlabel("Clase (Pclass)")
plt.ylabel("Cantidad de pasajeros")
plt.show()
```

## 7. Distribución de género dentro de cada clase
En todas las clases hay más hombres que mujeres, y la diferencia es más marcada en 3ª clase (347 hombres vs. 144 mujeres).

```python
sns.countplot(data=df, x="Pclass", hue="Sex", palette="Set2")
plt.title("Distribución de género por clase de pasajero")
plt.xlabel("Clase (Pclass)")
plt.ylabel("Cantidad de pasajeros")
plt.legend(title="Género")
plt.show()
```

## 8. Distribución de clase por puerto de embarque
Southampton embarcó pasajeros de las tres clases, con predominio de 3ª. En Queenstown casi todos eran de 3ª clase (72 de 77) y Cherbourg tuvo una proporción alta de 1ª clase (85 de 168).

```python
sns.countplot(data=df, x="Embarked", hue="Pclass", order=["C", "Q", "S"], palette="Set1")
plt.title("Clase de pasajero por puerto de embarque")
plt.xlabel("Puerto de embarque")
plt.ylabel("Cantidad de pasajeros")
plt.legend(title="Clase")
plt.show()
```

## 9. Conteo de pasajeros por género
Hay 577 hombres (≈ 65 %) y 314 mujeres (≈ 35 %).

```python
sns.countplot(data=df, x="Sex", hue="Sex", palette="pastel", legend=False)
plt.title("Cantidad de pasajeros por género")
plt.xlabel("Género")
plt.ylabel("Cantidad de pasajeros")
plt.show()
```

## 10. Valores máximo y mínimo de la edad
El pasajero más joven tiene 0.42 años (unos 5 meses) y el mayor 80 años.

```python
print(f"Edad mínima: {df['Age'].min()}")
print(f"Edad máxima: {df['Age'].max()}")
```

## 11. Distribución de edades con media, mediana y moda
Se marcan con líneas verticales la media (≈ 29.4), la mediana (28) y la moda (28). Que la mediana y la moda coincidan refleja el relleno de valores faltantes con la mediana; la media queda ligeramente a la derecha por la cola de pasajeros mayores.

```python
media, mediana, moda = df["Age"].mean(), df["Age"].median(), df["Age"].mode()[0]

sns.histplot(df["Age"], bins=30, kde=True, color="lightsteelblue")
plt.axvline(media, color="red", linestyle="--", label=f"Media: {media:.2f}")
plt.axvline(mediana, color="green", linestyle="-.", label=f"Mediana: {mediana:.2f}")
plt.axvline(moda, color="purple", linestyle=":", linewidth=2, label=f"Moda: {moda:.2f}")
plt.title("Distribución de edades con media, mediana y moda")
plt.xlabel("Edad")
plt.ylabel("Cantidad de pasajeros")
plt.legend()
plt.show()
```

## 12. Boxplot de edad (outliers)
Los puntos fuera de los bigotes son valores atípicos según el criterio del rango intercuartílico (1.5 × IQR). Aparecen outliers por encima (pasajeros de edad avanzada) y algunos por debajo (niños muy pequeños dada la concentración de edades en torno a 28 años).

```python
sns.boxplot(x=df["Age"], color="skyblue")
plt.title("Boxplot de edad")
plt.xlabel("Edad")
plt.show()

Q1, Q3 = df["Age"].quantile([0.25, 0.75])
IQR = Q3 - Q1
lim_inf, lim_sup = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
outliers_edad = df[(df["Age"] < lim_inf) | (df["Age"] > lim_sup)]
print(f"Límites: [{lim_inf:.2f}, {lim_sup:.2f}] -> {len(outliers_edad)} outliers")
```

## 13. Valores máximo y mínimo de la tarifa (Fare)
La tarifa mínima es 0 (pasajeros sin pago registrado) y la máxima ≈ 512.33, muy por encima del promedio.

```python
print(f"Tarifa mínima: {df['Fare'].min()}")
print(f"Tarifa máxima: {df['Fare'].max()}")
```

## 14. Distribución de tarifas con media, mediana y moda
La distribución tiene un fuerte sesgo a la derecha: la media (≈ 32.20) es más del doble que la mediana (14.45) y la moda es 8.05, porque unas pocas tarifas muy altas jalan el promedio hacia arriba.

```python
media_f, mediana_f, moda_f = df["Fare"].mean(), df["Fare"].median(), df["Fare"].mode()[0]

sns.histplot(df["Fare"], bins=40, kde=True, color="lightcoral")
plt.axvline(media_f, color="red", linestyle="--", label=f"Media: {media_f:.2f}")
plt.axvline(mediana_f, color="green", linestyle="-.", label=f"Mediana: {mediana_f:.2f}")
plt.axvline(moda_f, color="purple", linestyle=":", linewidth=2, label=f"Moda: {moda_f:.2f}")
plt.title("Distribución de tarifas con media, mediana y moda")
plt.xlabel("Tarifa (Fare)")
plt.ylabel("Cantidad de pasajeros")
plt.legend()
plt.show()
```

## 15. Boxplot de tarifa (outliers)
Con el criterio 1.5 × IQR hay 116 tarifas atípicas, todas por encima del límite superior (≈ 65.6). Son tarifas de pasajes caros, principalmente de 1ª clase, y no necesariamente errores de datos.

```python
sns.boxplot(x=df["Fare"], color="salmon")
plt.title("Boxplot de tarifa")
plt.xlabel("Tarifa (Fare)")
plt.show()

Q1, Q3 = df["Fare"].quantile([0.25, 0.75])
IQR = Q3 - Q1
lim_sup = Q3 + 1.5 * IQR
outliers_fare = df[df["Fare"] > lim_sup]
print(f"Límite superior: {lim_sup:.2f} -> {len(outliers_fare)} outliers")
```

## 16. Pasajeros que pagaron 0 de tarifa
Hay 15 pasajeros con tarifa 0, todos hombres que embarcaron en Southampton y viajaban solos. Solo uno sobrevivió. Pueden ser tripulantes, invitados o datos no registrados.

```python
tarifa_cero = df[df["Fare"] == 0]
print(f"Pasajeros con tarifa 0: {len(tarifa_cero)}")
tarifa_cero
```

## 17. Conteo de pasajeros por hermanos/esposos a bordo (SibSp)
La mayoría viajaba sin hermanos ni cónyuge (SibSp = 0, 608 pasajeros). Los valores altos (5 u 8) son muy poco frecuentes.

```python
sns.countplot(data=df, x="SibSp", hue="SibSp", palette="mako", legend=False)
plt.title("Pasajeros por número de hermanos/esposos a bordo")
plt.xlabel("SibSp")
plt.ylabel("Cantidad de pasajeros")
plt.show()
```

## 18. Tasa de supervivencia según SibSp
Quienes viajaban con 1 o 2 familiares (hermano/esposo) tuvieron mayor tasa de supervivencia que quienes viajaban solos; con 5 o más la supervivencia fue nula. Con pocos pasajeros en los valores altos, esas tasas son poco robustas.

```python
tasa_sibsp = df.groupby("SibSp")["Survived"].mean().reset_index()

sns.barplot(data=tasa_sibsp, x="SibSp", y="Survived", hue="SibSp", palette="rocket", legend=False)
plt.title("Tasa de supervivencia por SibSp")
plt.xlabel("SibSp")
plt.ylabel("Tasa de supervivencia")
plt.ylim(0, 1)
plt.show()
```

## 19. Porcentaje de supervivencia para cada valor de SibSp
Se calcula el promedio de `Survived` por valor de `SibSp` y se multiplica por 100.

```python
porcentaje_sibsp = (df.groupby("SibSp")["Survived"].mean() * 100).round(2).rename("Supervivencia (%)")
porcentaje_sibsp.to_frame()
```

## 20. Mapa de calor de la matriz de correlación
Correlación entre variables numéricas. Las relaciones más notables: `Pclass` con `Fare` (≈ -0.55, a mejor clase, tarifa más alta), `Survived` con `Pclass` (≈ -0.34) y con `Fare` (≈ 0.26), y `SibSp` con `Parch` (≈ 0.41). Ninguna correlación con `Survived` es fuerte por sí sola.

```python
corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, square=True)
plt.title("Matriz de correlación")
plt.show()
```
